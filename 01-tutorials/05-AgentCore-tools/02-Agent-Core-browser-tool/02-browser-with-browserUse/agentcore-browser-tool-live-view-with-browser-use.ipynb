## 🔧 Environment Setup
Set up the advanced environment for live view browser automation.

```
import os
import asyncio
import json
import time
from typing import Dict, Any, Optional
from datetime import datetime
import boto3
from botocore.exceptions import ClientError

# Import AgentCore Browser Tool with Live View capabilities
from amazon_bedrock_agentcore.browser_tool import BrowserTool
from amazon_bedrock_agentcore.browser_tool.models import BrowserAction
from amazon_bedrock_agentcore.live_view import LiveViewManager

# Import Browser-Use SDK
from browser_use import Agent, Browser
from browser_use.browser.browser import BrowserConfig
from browser_use.agent.service import AgentService

print("✅ All imports successful!")
```

## 🔑 Configuration Setup
Configure the environment for live view browser automation.

```
# AWS Configuration
AWS_REGION = "us-east-1"
AWS_PROFILE = None

# Live View Configuration
LIVE_VIEW_ENABLED = True
LIVE_VIEW_PORT = 8080
LIVE_VIEW_RESOLUTION = {"width": 1920, "height": 1080}
LIVE_VIEW_QUALITY = "high"

# Browser-Use Configuration
BROWSER_USE_MODEL = "anthropic.claude-3-5-sonnet-20241022-v2:0"
BROWSER_USE_HEADLESS = False  # Enable visual mode for live view

# Initialize AWS session
try:
    if AWS_PROFILE:
        session = boto3.Session(profile_name=AWS_PROFILE, region_name=AWS_REGION)
    else:
        session = boto3.Session(region_name=AWS_REGION)
    
    # Test AWS credentials
    sts = session.client('sts')
    identity = sts.get_caller_identity()
    print(f"✅ AWS credentials configured for account: {identity['Account']}")
    
    # Verify Bedrock access
    bedrock = session.client('bedrock-runtime')
    print("✅ Amazon Bedrock access verified")
    
except ClientError as e:
    print(f"❌ AWS configuration error: {e}")
```

## 🎥 Initialize Live View System
Set up the live view system with Browser-Use integration.

```
class LiveViewBrowserAutomation:
    """Advanced browser automation with live view capabilities."""
    
    def __init__(self, aws_session: boto3.Session):
        """Initialize the live view automation system."""
        self.aws_session = aws_session
        self.browser_tool = None
        self.browser_use_agent = None
        self.live_view_manager = None
        self.browser_config = None
        self.session_active = False
        
    async def initialize_live_system(self):
        """Initialize all live view components."""
        try:
            print("🚀 Initializing live view browser automation system...")
            
            # Initialize AgentCore Browser Tool with Live View
            self.browser_tool = BrowserTool(
                session=self.aws_session,
                headless=BROWSER_USE_HEADLESS,
                viewport_size=LIVE_VIEW_RESOLUTION,
                live_view_enabled=LIVE_VIEW_ENABLED,
                live_view_port=LIVE_VIEW_PORT
            )
            
            # Initialize Live View Manager
            self.live_view_manager = LiveViewManager(
                browser_tool=self.browser_tool,
                port=LIVE_VIEW_PORT,
                quality=LIVE_VIEW_QUALITY,
                enable_interaction=True
            )
            
            # Configure Browser-Use
            self.browser_config = BrowserConfig(
                headless=BROWSER_USE_HEADLESS,
                viewport_size=LIVE_VIEW_RESOLUTION,
                chrome_instance_path=None  # Use system Chrome
            )
            
            # Initialize Browser-Use Agent
            self.browser_use_agent = Agent(
                task="Browser automation with live view",
                llm=self.aws_session.client('bedrock-runtime'),
                browser=Browser(config=self.browser_config),
                use_vision=True,
                save_conversation_path="./browser_use_conversations/"
            )
            
            print("✅ AgentCore Browser Tool with Live View initialized")
            print("✅ Live View Manager initialized")
            print("✅ Browser-Use Agent initialized")
            
            return True
            
        except Exception as e:
            print(f"❌ Live system initialization error: {e}")
            return False
    
    async def start_live_session(self):
        """Start the live automation session."""
        try:
            # Start browser with live view
            await self.browser_tool.start()
            print("✅ Browser session started")
            
            # Start live view streaming
            await self.live_view_manager.start_streaming()
            print(f"✅ Live view streaming started on port {LIVE_VIEW_PORT}")
            print(f"🌐 Access live view at: http://localhost:{LIVE_VIEW_PORT}")
            
            # Initialize Browser-Use browser
            await self.browser_use_agent.browser.start()
            print("✅ Browser-Use browser initialized")
            
            self.session_active = True
            return True
            
        except Exception as e:
            print(f"❌ Live session startup error: {e}")
            return False
    
    async def execute_with_live_monitoring(self, task: str, **kwargs):
        """Execute Browser-Use task with live monitoring."""
        try:
            print(f"🎬 Executing with live monitoring: {task}")
            
            # Start live recording
            recording_id = f"browser_use_{int(time.time())}"
            await self.live_view_manager.start_recording(recording_id)
            
            # Execute Browser-Use task
            result = await self.browser_use_agent.run(task)
            
            # Stop recording
            recording_path = await self.live_view_manager.stop_recording()
            
            print(f"✅ Live execution completed")
            print(f"📹 Recording saved: {recording_path}")
            print(f"📊 Task result: {result}")
            
            return result
            
        except Exception as e:
            print(f"❌ Live monitoring execution error: {e}")
            return None
    
    async def take_manual_control(self):
        """Allow manual control of the browser session."""
        try:
            print("🎛️ Enabling manual control...")
            await self.live_view_manager.enable_manual_control()
            print("✅ Manual control enabled")
            print("🌐 You can now interact with the browser through the live view")
            return True
        except Exception as e:
            print(f"❌ Manual control error: {e}")
            return False
    
    async def release_manual_control(self):
        """Release manual control back to automation."""
        try:
            print("🤖 Releasing manual control back to automation...")
            await self.live_view_manager.disable_manual_control()
            print("✅ Automation control restored")
            return True
        except Exception as e:
            print(f"❌ Control release error: {e}")
            return False

# Initialize the live automation system
live_automation = LiveViewBrowserAutomation(session)
```

## 🚀 Start Live View Automation
Launch the live view automation environment.

```
async def setup_live_view_automation():
    """Set up the complete live view automation environment."""
    
    print("🎬 Setting up live view browser automation...")
    
    # Initialize live system
    if not await live_automation.initialize_live_system():
        return False
    
    # Start live session
    if not await live_automation.start_live_session():
        return False
    
    print("🎉 Live view automation environment ready!")
    print(f"🌐 Live view available at: http://localhost:{LIVE_VIEW_PORT}")
    print("👀 You can now watch Browser-Use agents work in real-time!")
    
    return True

# Run the setup
if asyncio.get_event_loop().is_running():
    import nest_asyncio
    nest_asyncio.apply()

live_setup_success = await setup_live_view_automation()

if live_setup_success:
    print("✅ Ready for live view browser automation!")
    print("🎥 Open your browser and navigate to the live view URL to watch!")
else:
    print("❌ Live setup failed. Please check your configuration.")
```

## 🎯 Basic Live Automation Example
Watch Browser-Use perform automation tasks in real-time.

```
async def live_automation_example():
    """Example: Basic automation with live visualization."""
    
    if not live_setup_success:
        print("⚠️  Skipping - live setup not completed")
        return None
    
    try:
        print("🎬 Starting live automation example...")
        print("👀 Watch the live view to see Browser-Use in action!")
        
        # Simple navigation task
        task = "Navigate to https://example.com and take a screenshot of the page"
        
        result = await live_automation.execute_with_live_monitoring(task)
        
        if result:
            print("✅ Live automation example completed successfully")
            print(f"📊 Result: {result}")
        else:
            print("❌ Live automation example failed")
        
        return result
        
    except Exception as e:
        print(f"❌ Live automation error: {e}")
        return None

# Execute the live automation example
if live_setup_success:
    live_result = await live_automation_example()
else:
    print("⚠️  Skipping live automation example")
```

## 🎛️ Interactive Control Example
Demonstrate manual control capabilities during automation.

```
async def interactive_control_example():
    """Example: Interactive control during automation."""
    
    if not live_setup_success:
        print("⚠️  Skipping - live setup not completed")
        return None
    
    try:
        print("🎛️ Starting interactive control example...")
        
        # Start an automation task
        task = "Navigate to a news website and find the top headline"
        
        print("🤖 Starting automated task...")
        print("🎬 You can take manual control at any time through the live view")
        
        # Start the task (this will run in background)
        automation_task = asyncio.create_task(
            live_automation.execute_with_live_monitoring(task)
        )
        
        # Wait a bit, then demonstrate manual control
        await asyncio.sleep(5)
        
        print("\n🎛️ Demonstrating manual control...")
        await live_automation.take_manual_control()
        
        print("⏳ Manual control active for 10 seconds...")
        print("🌐 You can now interact with the browser manually")
        await asyncio.sleep(10)
        
        print("\n🤖 Releasing control back to automation...")
        await live_automation.release_manual_control()
        
        # Wait for automation to complete
        result = await automation_task
        
        if result:
            print("✅ Interactive control example completed")
        
        return result
        
    except Exception as e:
        print(f"❌ Interactive control error: {e}")
        return None

# Execute interactive control example
if live_setup_success:
    print("\n" + "="*60)
    print("🎛️ INTERACTIVE CONTROL DEMONSTRATION")
    print("="*60)
    interactive_result = await interactive_control_example()
else:
    print("⚠️  Skipping interactive control example")
```

## 📊 Session Monitoring and Analytics
Monitor automation performance and gather analytics.

```
async def session_analytics_example():
    """Example: Gather session analytics and monitoring data."""
    
    if not live_setup_success:
        print("⚠️  Skipping - live setup not completed")
        return None
    
    try:
        print("📊 Starting session analytics collection...")
        
        # Start analytics collection
        analytics_session = await live_automation.live_view_manager.start_analytics()
        
        # Perform automation with analytics
        analytics_task = "Navigate to multiple websites and collect their titles"
        
        print("🎬 Executing automation with analytics collection...")
        
        result = await live_automation.execute_with_live_monitoring(
            analytics_task,
            collect_analytics=True
        )
        
        # Stop analytics and get results
        analytics_data = await live_automation.live_view_manager.stop_analytics()
        
        if result and analytics_data:
            print("✅ Session analytics completed")
            
            # Display analytics
            print("\n📈 Session Analytics:")
            print(f"   Session duration: {analytics_data.get('duration', 'N/A')} seconds")
            print(f"   Pages visited: {analytics_data.get('pages_visited', 'N/A')}")
            print(f"   Actions performed: {analytics_data.get('actions_count', 'N/A')}")
            print(f"   Screenshots taken: {analytics_data.get('screenshots', 'N/A')}")
            
            # Live view statistics
            print("\n📺 Live View Statistics:")
            print(f"   Streaming quality: {analytics_data.get('stream_quality', 'N/A')}")
            print(f"   Average FPS: {analytics_data.get('fps', 'N/A')}")
            print(f"   Bandwidth used: {analytics_data.get('bandwidth_mb', 'N/A')} MB")
        
        return analytics_data
        
    except Exception as e:
        print(f"❌ Session analytics error: {e}")
        return None

# Execute session analytics
if live_setup_success:
    print("\n" + "="*60)
    print("📊 SESSION ANALYTICS COLLECTION")
    print("="*60)
    analytics_data = await session_analytics_example()
else:
    print("⚠️  Skipping session analytics example")
```

## 🧹 Cleanup and Session Management
Properly clean up resources and save session data.

```
async def cleanup_live_session():
    """Clean up live session resources."""
    
    try:
        print("🧹 Starting live session cleanup...")
        
        if live_automation.session_active:
            # Stop live view streaming
            if live_automation.live_view_manager:
                await live_automation.live_view_manager.stop_streaming()
                print("✅ Live view streaming stopped")
            
            # Close Browser-Use browser
            if live_automation.browser_use_agent and live_automation.browser_use_agent.browser:
                await live_automation.browser_use_agent.browser.close()
                print("✅ Browser-Use browser closed")
            
            # Close AgentCore browser
            if live_automation.browser_tool:
                await live_automation.browser_tool.close()
                print("✅ AgentCore browser session closed")
            
            # Save session summary
            session_summary = {
                "timestamp": datetime.now().isoformat(),
                "live_view_port": LIVE_VIEW_PORT,
                "resolution": LIVE_VIEW_RESOLUTION,
                "session_active": live_automation.session_active
            }
            
            session_file = f"live_session_{int(time.time())}.json"
            with open(session_file, 'w') as f:
                json.dump(session_summary, f, indent=2)
            print(f"💾 Session summary saved to: {session_file}")
        
        print("✅ Live session cleanup completed")
        
    except Exception as e:
        print(f"⚠️  Cleanup error: {e}")

# Perform cleanup
await cleanup_live_session()
```

## 🔧 Environment Setup
Let's start by setting up our environment and importing the necessary libraries.

```
import os
import asyncio
import json
from typing import Dict, Any, Optional
import boto3
from botocore.exceptions import ClientError

# Import AgentCore Browser Tool
from amazon_bedrock_agentcore.browser_tool import BrowserTool
from amazon_bedrock_agentcore.browser_tool.models import BrowserAction

# Import Nova Act SDK
from nova_act import NovaActClient, NovaActConfig

print("✅ All imports successful!")
```

## 🔑 Configuration Setup
Configure your AWS credentials and Nova Act API key.

```
# AWS Configuration
AWS_REGION = "us-east-1"  # Change to your preferred region
AWS_PROFILE = None  # Set to your AWS profile name if using profiles

# Nova Act Configuration
NOVA_ACT_API_KEY = os.getenv("NOVA_ACT_API_KEY")
if not NOVA_ACT_API_KEY:
    print("⚠️  Please set your NOVA_ACT_API_KEY environment variable")
    print("   export NOVA_ACT_API_KEY='your-api-key-here'")
else:
    print("✅ Nova Act API key configured")

# Initialize AWS session
try:
    if AWS_PROFILE:
        session = boto3.Session(profile_name=AWS_PROFILE, region_name=AWS_REGION)
    else:
        session = boto3.Session(region_name=AWS_REGION)
    
    # Test AWS credentials
    sts = session.client('sts')
    identity = sts.get_caller_identity()
    print(f"✅ AWS credentials configured for account: {identity['Account']}")
    
except ClientError as e:
    print(f"❌ AWS credentials error: {e}")
    print("Please configure your AWS credentials using:")
    print("  - AWS CLI: aws configure")
    print("  - Environment variables: AWS_ACCESS_KEY_ID, AWS_SECRET_ACCESS_KEY")
    print("  - IAM roles (if running on EC2)")
```

## 🚀 Initialize AgentCore Browser Tool
Set up the AgentCore Browser Tool for headless browser automation.

```
class NovaActBrowserAutomation:
    def __init__(self, aws_session: boto3.Session, nova_act_api_key: str):
        """Initialize the Nova Act browser automation system."""
        self.aws_session = aws_session
        self.nova_act_api_key = nova_act_api_key
        self.browser_tool = None
        self.nova_act_client = None
        
    async def initialize(self):
        """Initialize both AgentCore Browser Tool and Nova Act client."""
        try:
            # Initialize AgentCore Browser Tool
            self.browser_tool = BrowserTool(
                session=self.aws_session,
                headless=True,  # Run in headless mode
                viewport_size={"width": 1920, "height": 1080}
            )
            
            # Initialize Nova Act client
            nova_config = NovaActConfig(
                api_key=self.nova_act_api_key,
                timeout=30,
                max_retries=3
            )
            self.nova_act_client = NovaActClient(config=nova_config)
            
            print("✅ AgentCore Browser Tool initialized")
            print("✅ Nova Act client initialized")
            
            return True
            
        except Exception as e:
            print(f"❌ Initialization error: {e}")
            return False
    
    async def start_browser_session(self):
        """Start a new browser session."""
        try:
            await self.browser_tool.start()
            print("✅ Browser session started")
            return True
        except Exception as e:
            print(f"❌ Browser session error: {e}")
            return False
    
    async def close_browser_session(self):
        """Close the browser session."""
        try:
            await self.browser_tool.close()
            print("✅ Browser session closed")
        except Exception as e:
            print(f"❌ Error closing browser: {e}")

# Initialize the automation system
automation = NovaActBrowserAutomation(session, NOVA_ACT_API_KEY)
```

## 🔗 Establish Browser Connection
Connect to the browser and prepare for automation tasks.

```
async def setup_browser_automation():
    """Set up the complete browser automation environment."""
    
    # Initialize the automation system
    if not await automation.initialize():
        return False
    
    # Start browser session
    if not await automation.start_browser_session():
        return False
    
    print("🎉 Browser automation environment ready!")
    return True

# Run the setup
if asyncio.get_event_loop().is_running():
    # If running in Jupyter, create a new event loop
    import nest_asyncio
    nest_asyncio.apply()

setup_success = await setup_browser_automation()

if setup_success:
    print("✅ Ready to proceed with Nova Act automation examples!")
else:
    print("❌ Setup failed. Please check your configuration.")
```

## 🎯 Basic Nova Act Automation Example
Let's start with a simple example of using Nova Act for web automation.

```
async def nova_act_navigate_example():
    """Example: Navigate to a website using Nova Act natural language commands."""
    
    try:
        # Natural language instruction for Nova Act
        instruction = "Navigate to https://example.com and take a screenshot"
        
        # Use Nova Act to interpret and execute the instruction
        result = await automation.nova_act_client.execute_instruction(
            instruction=instruction,
            browser_context=automation.browser_tool.get_context()
        )
        
        print(f"✅ Nova Act execution result: {result.status}")
        print(f"📝 Action taken: {result.action_description}")
        
        if result.screenshot:
            print("📸 Screenshot captured successfully")
            
        return result
        
    except Exception as e:
        print(f"❌ Nova Act execution error: {e}")
        return None

# Execute the example
if setup_success:
    navigation_result = await nova_act_navigate_example()
else:
    print("⚠️  Skipping example - setup not completed")
```

## 🔍 Advanced Data Extraction Example
Use Nova Act for intelligent data extraction from web pages.

```
async def nova_act_data_extraction_example():
    """Example: Extract data from a webpage using natural language."""
    
    try:
        # Navigate to a data-rich page
        navigation_instruction = "Go to https://news.ycombinator.com"
        nav_result = await automation.nova_act_client.execute_instruction(
            instruction=navigation_instruction,
            browser_context=automation.browser_tool.get_context()
        )
        
        if nav_result.status == "success":
            print("✅ Successfully navigated to Hacker News")
            
            # Extract data using natural language
            extraction_instruction = """
            Extract the top 5 news headlines from this page. 
            For each headline, get the title, points, and number of comments.
            Return the data in JSON format.
            """
            
            extraction_result = await automation.nova_act_client.execute_instruction(
                instruction=extraction_instruction,
                browser_context=automation.browser_tool.get_context(),
                return_format="json"
            )
            
            if extraction_result.status == "success":
                print("✅ Data extraction completed")
                print("📊 Extracted data:")
                
                # Parse and display the extracted data
                if extraction_result.data:
                    extracted_data = json.loads(extraction_result.data)
                    for i, item in enumerate(extracted_data.get("headlines", []), 1):
                        print(f"{i}. {item.get('title', 'N/A')}")
                        print(f"   Points: {item.get('points', 'N/A')}")
                        print(f"   Comments: {item.get('comments', 'N/A')}")
                        print()
                
                return extraction_result
            else:
                print(f"❌ Data extraction failed: {extraction_result.error}")
                
        else:
            print(f"❌ Navigation failed: {nav_result.error}")
            
    except Exception as e:
        print(f"❌ Data extraction error: {e}")
        return None

# Execute the data extraction example
if setup_success:
    extraction_result = await nova_act_data_extraction_example()
else:
    print("⚠️  Skipping example - setup not completed")
```

## 🛠️ Error Handling and Best Practices
Implement robust error handling for production automation.

```
class RobustNovaActAutomation:
    """Enhanced automation class with comprehensive error handling."""
    
    def __init__(self, automation_instance):
        self.automation = automation_instance
        self.retry_count = 3
        self.timeout = 30
    
    async def execute_with_retry(self, instruction: str, **kwargs):
        """Execute Nova Act instruction with retry logic."""
        
        for attempt in range(self.retry_count):
            try:
                print(f"🔄 Attempt {attempt + 1}/{self.retry_count}: {instruction[:50]}...")
                
                result = await self.automation.nova_act_client.execute_instruction(
                    instruction=instruction,
                    browser_context=self.automation.browser_tool.get_context(),
                    timeout=self.timeout,
                    **kwargs
                )
                
                if result.status == "success":
                    print(f"✅ Success on attempt {attempt + 1}")
                    return result
                else:
                    print(f"⚠️  Attempt {attempt + 1} failed: {result.error}")
                    
            except Exception as e:
                print(f"❌ Attempt {attempt + 1} error: {e}")
                
                if attempt < self.retry_count - 1:
                    print(f"🔄 Retrying in 2 seconds...")
                    await asyncio.sleep(2)
        
        print(f"❌ All {self.retry_count} attempts failed")
        return None
    
    async def safe_cleanup(self):
        """Safely clean up resources."""
        try:
            await self.automation.close_browser_session()
            print("✅ Cleanup completed successfully")
        except Exception as e:
            print(f"⚠️  Cleanup warning: {e}")

# Create robust automation instance
if setup_success:
    robust_automation = RobustNovaActAutomation(automation)
    
    # Example with retry logic
    test_instruction = "Take a screenshot of the current page"
    robust_result = await robust_automation.execute_with_retry(test_instruction)
    
    if robust_result:
        print("🎉 Robust execution successful!")
else:
    print("⚠️  Skipping robust automation example")
```

## 🧹 Cleanup and Session Management
Properly clean up resources and close browser sessions.

```
async def cleanup_automation():
    """Clean up all automation resources."""
    
    try:
        if 'robust_automation' in locals():
            await robust_automation.safe_cleanup()
        elif 'automation' in locals():
            await automation.close_browser_session()
        
        print("✅ All resources cleaned up successfully")
        
    except Exception as e:
        print(f"⚠️  Cleanup error: {e}")

# Perform cleanup
await cleanup_automation()
```
